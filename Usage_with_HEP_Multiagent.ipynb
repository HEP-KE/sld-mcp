# Demo: Using `sld-mcp` with `HEP-multiagent`

This notebook demonstrates how to connect the `HEP-multiagent` framework to the
`SLD MCP server` for public SLD data analysis.

## Prerequisites

This notebook assumes:

- Local access to the [public SLD parquet dataset](https://zenodo.org/records/21199778) (or run `sld-resurrect download-dataset`)
- A working installation of [SLD Resurrection](https://github.com/HEP-KE/sld-resurrect)
- A working installation of [HEP-multiagent](https://github.com/HEP-KE/HEP-multiagent)
- Python 3.10+

**Install commands:**

```bash
# SLD Resurrection
git clone https://github.com/HEP-KE/sld-resurrect.git
cd sld-resurrect
pip install -e .

# HEP-multiagent
pip install git+https://github.com/HEP-KE/HEP-multiagent.git

## Setup

Here is how to configure the HEP-Multiagent with the SLD MCP server.

````python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from hep_multiagent import Agent

load_dotenv()


llm = ChatOpenAI(
    model="gpt-4",
    api_key=os.environ.get("OPENAI_API_KEY", "")
)


agent = await Agent(
    llm=llm,
    mcp_servers=[{
        "url": "https://github.com/HEP-KE/sld-mcp.git",
        "name": "sld_mcp_server",
    }],
    approval=False,
)
````

## Example Query

````python
result = await agent.run(
    query=(
        "Using the local SLD parquet files, create plots of the polar angle "
        "distribution separated by beam helicity for the three leptonic channels "
        "(electron, muon, and tau). Save the plots to the output directory and "
        "briefly summarize what was produced."
    ),
    output_dir="./sld_mcp_demo_output",
)
````
